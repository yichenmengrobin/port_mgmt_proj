```python
from pathlib import Path

import pandas as pd
from IPython.display import display
```

```python
data_path = Path("data/XRD_COV_MATRIX.xlsx")
if not data_path.is_file():
    data_path = Path("../data/XRD_COV_MATRIX.xlsx")
raw = pd.read_excel(data_path, header=None, skiprows=2)
xrd_mat = raw.iloc[:, 3:].copy()
xrd_mat.index = raw.iloc[:, 1].to_numpy()
xrd_mat.columns = xrd_mat.index.copy()
xrd_mat.index.name = "Ticker"
xrd_mat.columns.name = "Ticker"

with pd.option_context("display.max_rows", None, "display.max_columns", None):
    display(xrd_mat)
```

Expected returns are annualized geometric time-weighted returns: $\mu_i = [\prod_{t=1}^{n_i}(1+r_{i,t})]^{13/n_i}-1$. There are 13 four-week periods per year. Assuming XRD is four-week covariance, annual covariance is `13 * xrd_mat`. Weights sum to 100%, with a minimum holding of 0% and a risk-free rate of zero.

```python
returns = pd.read_csv(data_path.parent / "stagger_3_returns.csv", index_col="date")
r = returns[xrd_mat.index]
mu = (1 + r).prod() ** (13 / r.count()) - 1
annual_cov = 13 * xrd_mat  # Assumes XRD contains four-week covariance.

display(mu.rename("Annualized geometric return"))
```

```python
from pypfopt import EfficientFrontier, plotting
import matplotlib.pyplot as plt

# Long-only, fully invested; the 100% upper bound is implied by these constraints.
ef = EfficientFrontier(mu, annual_cov, weight_bounds=(0, 1))
plotting.plot_efficient_frontier(ef.deepcopy())
ef.max_sharpe(risk_free_rate=0)
ret, vol, sharpe = ef.portfolio_performance(risk_free_rate=0)
plt.scatter(vol, ret, marker="*", s=200, label="Maximum Sharpe")
plt.xlabel("Annualized volatility (assuming four-week XRD)")
plt.ylabel("Expected annualized return")
plt.legend()
plt.show()

display(pd.Series(ef.clean_weights(), name="Weight"))
display(pd.Series({"Annualized return": ret, "Annualized volatility": vol, "Sharpe (rf=0)": sharpe}))
```

```python
# Add bounds only when a holding violates a limit; keep previous bounds.
lower = pd.Series(0.0, index=mu.index)
upper = pd.Series(1.0, index=mu.index)
weights = pd.Series(ef.weights, index=mu.index)
runs = []

for run in range(2 * len(mu) + 1):
    below = weights < 0.001 - 1e-7
    above = weights > 0.10 + 1e-7
    if not (below.any() or above.any()):
        break
    lower[below] = 0.001
    upper[above] = 0.10
    constrained_ef = EfficientFrontier(mu, annual_cov, weight_bounds=list(zip(lower, upper)))
    constrained_ef.max_sharpe(risk_free_rate=0)
    weights = pd.Series(constrained_ef.weights, index=mu.index)
    ret, vol, sharpe = constrained_ef.portfolio_performance(risk_free_rate=0)
    runs.append({"Rerun": run + 1, "Minimum weight": weights.min(),
                 "Maximum weight": weights.max(), "Sharpe (rf=0)": sharpe})
else:
    raise RuntimeError("Holding limits did not converge")

if runs:
    display(pd.DataFrame(runs).set_index("Rerun"))
    plotting.plot_efficient_frontier(
        EfficientFrontier(mu, annual_cov, weight_bounds=list(zip(lower, upper)))
    )
    plt.scatter(vol, ret, marker="*", s=200, label="Maximum Sharpe")
    plt.title("Efficient frontier with final holding constraints")
    plt.xlabel("Annualized volatility (assuming four-week XRD)")
    plt.ylabel("Expected annualized return")
    plt.legend()
    plt.show()
    display(pd.Series({"Annualized return": ret, "Annualized volatility": vol,
                       "Sharpe (rf=0)": sharpe}))
else:
    print("First-run holdings already meet the limits; no rerun needed.")

display(weights.rename("Final weight").to_frame().style.format("{:.3%}"))
```
